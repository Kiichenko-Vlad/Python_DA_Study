8. Для продовження дослідження різниці у прибутках в магазинах давайте зʼясуємо, чи відрізняється маржа ('gross margin percentage') для різних інвойсів в магазині? Перевірте це програмно на даних виходячи з того функціоналу, який ми вже пройшли у лекціях.

Дайте на основі отриманого результату відповідь: "У всіх всіх інвойсах однакова маржа" або "Маржа відрізняється для різних інвойсів".

```python
for city in df['City'].unique():
    margins = df[df['City'] == city]['gross margin percentage'].unique()
    print(f"{city}: {margins} (кількість унікальних: {len(margins)})")
```

```python
unique_margins = df['gross margin percentage'].nunique()

if unique_margins == 1:
  print('У всіх інвойсах однакова маржа')
else:
  print('Маржа відрізняється для різних інвойсів')
```

9. Врахуовуючи висновок в попередньому завданні, дослідимо як продаються в цих містах різні категорії товарів. Виведіть відсотковий розпоіділ кількості інвойсів по категоріям товарів `Product line` в містах `Yangon` та `Mandalay`.

```python
data_yangon = df.query('City == "Yangon"')
data_mandalay = df.query('City == "Mandalay"')

dist_per_yangon = data_yangon['Product line'].value_counts(normalize=True)
dist_per_mandalay = data_mandalay['Product line'].value_counts(normalize=True)

print("Розподіл по категоріям товарів Product line в місті Yangon:")
print(dist_per_yangon)
print('\n')
print("Розподіл по категоріям товарів Product line в місті Mandalay:")
print(dist_per_mandalay)
```

```python
cities = ['Yangon', 'Mandalay']

for city in cities:
    dist = df.query(f'City == "{city}"')['Product line'].value_counts(normalize=True)
    print(f"Розподіл по категоріям товарів Product line в місті {city}:")
    print(dist)
    print('\n')
```
