###Гіпотеза №1
Чи правда, що покупці, які залишають високий рейтинг, витрачають більше грошей, ніж ті, хто ставить низький рейтинг?

```python
def rating_mapping(val):
  if val >= 8:
    return 'High'
  elif 6 <= val < 8:
    return 'Medium'
  else:
    return 'Low'

df['Rating Category'] = df.Rating.apply(rating_mapping)
```

```python
df.head(5)
```

```python
df.groupby('Rating Category')['Total'].mean().sort_values(ascending=False).round(3)
```

Парадокс:
Незадоволені клієнти (низький рейтинг) витрачають БІЛЬШЕ грошей, ніж задоволені!
Це протилежно тому, що очікувалося отримати!

```python
# Перевірка скільки людей в кожній категорії:
df.groupby('Rating Category').size().sort_values(ascending=False)
```

Ключовий інсайт:
Близько третини клієнтів (315 людей) залишають низький рейтинг, але витрачають на 10% більше (340.57 vs 307.10), ніж задоволені клієнти. Це означає, що найцінніші клієнти незадоволені сервісом — критична проблема для бізнесу.

Бізнес-ризик:
Якщо ці 315 клієнтів не повернуться через незадоволеність, супермаркет втратить ~107,000 доходу (315 × 340.57), що становить понад 30% від загального обороту.

Рекомендація:
Терміново покращити досвід для клієнтів з великими чеками (>300) — швидші каси, VIP-зони, допомога з паковкою. Це утримає найприбутковіших клієнтів.

###Гіпотеза №2
Чи правда, що в обідній час (13:00-15:00) купують більше Food and beverages, ніж в інші години?

```python
def get_hour(time):
    return int(time.split(':')[0])

df['Hour'] = df.Time.apply(get_hour)
df.head()
```

```python
# Обідній час
lunch_food = df[(df['Hour'].isin([13, 14, 15])) & (df['Product line'] == 'Food and beverages')]

# Решта часу
non_lunch_food = df[(~df['Hour'].isin([13, 14, 15])) & (df['Product line'] == 'Food and beverages')]

print(f"Обід (13-15): {len(lunch_food)} інвойсів")
print(f"Не-обід: {len(non_lunch_food)} інвойсів")
```

```python
# Які категорії популярні в обід?
lunch_categories = df[df['Hour'].isin([13, 14, 15])]['Product line'].value_counts()
print(lunch_categories)
```

```python
# Кількість Food and beverages по кожній годині
food_by_hour = df[df['Product line'] == 'Food and beverages'].groupby('Hour').size().sort_values(ascending=False)
print(food_by_hour)
```

###Гіпотеза №3
Чи правда, що дрібні покупки (Total < 50) частіше відбуваються в певних категоріях товарів (наприклад, Food and beverages або Health and beauty), ніж великі покупки (Total > 500)?

```python
# Дрібні покупки
small_purchases = df[df['Total'] < 50]

# Великі покупки
large_purchases = df[df['Total'] > 500]

print(f"Дрібних покупок: {len(small_purchases)}")
print(f"Великих покупок: {len(large_purchases)}")
```

```python
# Топ категорії для дрібних покупок
print("=== Топ категорії для дрібних покупок (<50) ===")
small_categories = small_purchases['Product line'].value_counts()
print(small_categories)
print(f"\nВідсотки:")
print((small_categories / len(small_purchases) * 100).round(3))
```

```python
# Топ категорії для великих покупок
print("\n=== Топ категорії для великих покупок (>500) ===")
large_categories = large_purchases['Product line'].value_counts()
print(large_categories)
print(f"\nВідсотки:")
print((large_categories / len(large_purchases) * 100).round(3))
```

```python
# Створення зведеної таблиці
comparison = pd.DataFrame({
    'Дрібні (<50)': small_purchases['Product line'].value_counts(),
    'Великі (>500)': large_purchases['Product line'].value_counts()
}).fillna(0).astype(int)

# Додавання відсотків
comparison['Дрібні %'] = (comparison['Дрібні (<50)'] / len(small_purchases) * 100).round(3)
comparison['Великі %'] = (comparison['Великі (>500)'] / len(large_purchases) * 100).round(3)

# Сортування за дрібними покупками
comparison = comparison.sort_values('Дрібні (<50)', ascending=False)

print(comparison)
```

####Аналіз спростував очікування:

#####Великі покупки (>500) не концентруються на дорогих категоріях, а рівномірно розподілені по всіх категоріях (16-17% кожна), що свідчить про комплексні сімейні закупівлі.

#####Натомість дрібні покупки (<50) тяжіють до Fashion accessories (23%) як імпульсні покупки, тоді як Food and beverages найменш популярна для дрібних чеків (11.6%).

#####Бізнесу варто залучати великих покупців через VIP-програми та комплексні знижки, а для дрібних — розміщувати Fashion accessories біля кас для стимулювання імпульсних покупок.
