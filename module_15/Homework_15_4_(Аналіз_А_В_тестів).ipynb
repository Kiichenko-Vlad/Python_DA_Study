1. Для початку, уявімо, що ми тільки плануємо проведення зазначеного А/B-тесту і хочемо зрозуміти, дані про скількох користувачів нам треба зібрати, аби досягнути відчутного ефекту. Відчутним ефектом ми вважатимемо збільшення утримання на 1% після внесення зміни. Обчисліть, скільки користувачів сумарно нам треба аби досягнути такого ефекту, якщо продакт менеджер нам повідомив, що базове утримання є 19%.

####Гіпотеза:
H₀: 𝑝 = 𝑝0               
H₁: 𝑝 ≠ 𝑝0

````python
# Розрахунок розміру ефекту на основі наших очікуваних показників
effect_size = sms.proportion_effectsize(0.20, 0.19)
effect_size
````

````python
# Розрахунок необхідного розміру вибірки
required_n = sms.NormalIndPower().solve_power(
    effect_size,
    power=0.8,
    alpha=0.05,
    ratio=1
    )

required_n = ceil(required_n) * 2

print(required_n)
````

````python
control_results = df[df['version'] == 'gate_30']['retention_7']
treatment_results = df[df['version'] == 'gate_40']['retention_7']

n_con = control_results.count()
n_treat = treatment_results.count()
successes = [control_results.sum(), treatment_results.sum()]
nobs = [n_con, n_treat]

print(n_con)
print(n_treat)
successes
````

````python
z_stat, pval = proportions_ztest(successes, nobs=nobs)
(lower_con, lower_treat), (upper_con, upper_treat) = proportion_confint(successes, nobs=nobs, alpha=0.05)

print(f'z statistic: {z_stat:.2f}')
print(f'p-value: {pval:.3f}')
print(f'Довірчий інтервал 95% для групи control: [{lower_con:.3f}, {upper_con:.3f}]')
print(f'Довірчий інтервал 95% для групи treatment: [{lower_treat:.3f}, {upper_treat:.3f}]')
````

1. Чи є статистична значущою різниця між поведінкою користувачів у різних версіях гри?     
Відповідь:
      Так, різниця між поведінкою користувачів у різних версіях гри є статистична значущою. Оскільки наше p-value = 0.002 є меншим за рівень значущості (𝛼  = 0.05) ми можемо відхилити нульову гіпотезу та зробити висновок, що група gate_30 показує статистично значуще вище retention_7, ніж gate_40 на 7 день після встановлення застосунку. Тому варто залишити ворота на 30-му рівні.
2. Чи перетинаються довірчі інтервали утримання користувачів з різних версій гри? Про що це каже?      
Відповідь:
    Як бачимо, що довірчі інтервали в нас не перетинаються, тобто це дає нам візуальне підтвердження що різницями між групами є. Не перетинання довірчих інтервалів означає, що з 95% впевненістю істинні значення retention кожної групи лежать у різних діапазонах — тобто різниця між групами є реальною, а не випадковою.

4. Виконайте тест Хі-квадрат на рівні значущості 5% аби визначити, чи є залежність між версією гри та утриманням гравця на 7ий день після реєстрації.

    - Напишіть, як для цього тесту будуть сформульовані гіпотези.
    - Проведіть обчислення, виведіть p-значення і напишіть висновок за результатами тесту.


````python
H₀: версія гри та retention_7 є незалежними (немає залежності між змінними)
H₁: версія гри та retention_7 є залежними (залежність існує)
````

````python
table = pd.crosstab(df['version'], df['retention_7'])

chi2, p, dof, expected = stats.chi2_contingency(table)

print(f"χ² = {chi2:.3f}")
print(f"p-value = {p:.5f}")
print(f"Ступені свободи = {dof}")
print("Очікувані частоти:\n", expected)
````

Оскільки p-value = 0.0016 < 0.05, ми відхиляємо H₀ та робимо висновок, що між версією гри та retention_7 існує статистично значуща залежність.
