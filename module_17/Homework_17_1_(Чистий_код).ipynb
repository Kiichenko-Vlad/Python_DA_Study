````python
# Відредагований код:

def greet(name):  # тіло функції перенесено на новий рядок
    print(f"Hello, {name}")


def add(n1, n2):  # додано пробіли після ком у параметрах; тіло на новий рядок
    return n1 + n2  # додано пробіли навколо оператора +


x = 5  # розділено два оператори на окремі рядки (прибрано крапку з комою)
y = 3

print(greet("John"), add(x, y))  # display() → print(); додано пробіли після ком
````

**Виправлені порушення PEP8:**
- Тіло кожної функції перенесено на окремий рядок з відступом (замість запису в один рядок через двокрапку);
- Додано два порожніх рядки між визначеннями функцій;

**Виправлені порушення PEP8:**
- Назви класів `animal`, `dog` приведено до `CapWords` (PascalCase): `Animal`, `Dog`;
- `class animal: pass` — `pass` перенесено на окремий рядок з відступом;
- Додано два порожніх рядки між визначеннями класів;
- Додано пробіл після коми у `__init__(self, name)`;
- Додано пробіли навколо `=` при присвоєннях атрибутів.

5.У вас є DataFrame df, що представляє результати тестів студентів з математики, англійської мови та історії. Поточний код обчислює середній бал для кожного студента шляхом ітерації за рядками DataFrame за допомогою iterrows().

Ваше завдання полягає в тому, щоб змінити код так, аби уникнути використання iterrows() і замість цього застосувати більш ефективне та читабельне рішення відповідно до правил, озвучених у лекції. Рефакторинг треба зробити лише частині, де рахуються середні, імпорт і визначення датафрейму лишаємо без змін.

````python
import pandas as pd

df = pd.DataFrame({
    'student': ['Alice', 'Bob', 'Charlie', 'David'],
    'math_score': [89, 94, 77, 85],
    'english_score': [92, 88, 79, 85],
    'history_score': [84, 92, 81, 89]
})
````

````python
averages = []
for _, row in df.iterrows():
    averages.append((row['math_score'] + row['english_score'] + row['history_score']) / 3)

df['average_score'] = averages
````

````python
# Відредагований код — векторизована операція:
score_columns = ['math_score', 'english_score', 'history_score']
df['average_score'] = df[score_columns].mean(axis=1)

print(df)
````

**Пояснення:**  
`iterrows()` ітерує DataFrame рядок за рядком як Python-об'єкти, що дуже повільно на великих даних.  
Замість цього використано `.mean(axis=1)` — векторизована операція pandas, яка обчислює середнє по вказаних стовпцях для кожного рядка одночасно, без Python-циклу.  
Це швидше, лаконічніше і більш читабельно.

І «product_name», і «product_category» є рядками, що представляють численні, але обмежені унікальні назви продуктів і категорії відповідно.

Напишіть код для ефективного перетворення стовпців «product_name» і «product_category» на категорійний тип для ефективного використання пам’яті.
Спробуйте вирішити це завдання, мінімізуючи використання пам'яті та зберігаючи читабельність коду. Крім того, виміряйте використання пам’яті до та після перетворення, щоб продемонструвати підвищення ефективності.

Порівняйте кількість памʼяті, яку займає наданий набір даних і новий, створений Вами. В лекції був приклад. В скільки разів менше памʼяті займає новий датафрейм?

````python
import pandas as pd
import numpy as np

np.random.seed(0)

products = ['apple', 'banana', 'cherry', 'dates', 'elderberry']
categories = ['fruit', 'fruit', 'fruit', 'fruit', 'fruit']

N_ROWS = 10**6
df = pd.DataFrame({
    'product_name': np.random.choice(products, N_ROWS),
    'product_category': np.random.choice(categories, N_ROWS),
    'product_price': np.random.uniform(1, 10, N_ROWS),
    'transaction_date': pd.date_range(start='1/1/2000', periods=N_ROWS, freq='h')
})
````

````python
# Вимірювання пам'яті "до" конвертації
memory_before = df.memory_usage(deep=True).sum()
print(f"Пам'ять до конвертації: {memory_before / 1024**2:.2f} MB")

# Конвертація рядкових стовпців із обмеженою кількістю унікальних значень у category
df['product_name'] = df['product_name'].astype('category')
df['product_category'] = df['product_category'].astype('category')

# Вимірювання пам'яті "після" конвертації
memory_after = df.memory_usage(deep=True).sum()
print(f"Пам'ять після конвертації: {memory_after / 1024**2:.2f} MB")

# Порівняння
ratio = memory_before / memory_after
print(f"Новий DataFrame займає у {ratio:.1f} разів менше пам'яті")
````

**Пояснення:**  
Тип `object` (рядок) зберігає кожен рядок як окремий Python-об'єкт у пам'яті.  
Тип `category` зберігає лише унікальні значення один раз, а для кожного рядка — лише числовий індекс.  
Якщо унікальних значень мало (як тут: 5 продуктів і 1 категорія), економія пам'яті дуже суттєва.
